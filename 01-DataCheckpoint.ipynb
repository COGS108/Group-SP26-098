## Research Question

Does higher dietary linoleic acid intake, commonly found in fast-food and ultra-processed diets using seed oils, increase obesity prevalence and cardiovascular risk, measured by BMI and total cholesterol, among U.S. adults?

## Background and Prior Work

Dietary patterns in the United States have shifted toward greater consumption of ultra-processed foods, many of which are energy-dense and contain refined vegetable oils. Linoleic acid is an omega-6 polyunsaturated fatty acid commonly found in oils such as corn, sunflower, soybean, and canola oil.[^1] According to the CDC’s National Center for Health Statistics, ultra-processed foods made up about 55% of total daily calorie intake among people age 1 and older in the United States from August 2021 to August 2023, with an even higher percentage among youth.[^1] Prior research has also found that higher consumption of ultra-processed foods is associated with increased risk of obesity, type 2 diabetes, and cardiovascular disease.[^2] However, research on linoleic acid itself is more mixed, with some evidence suggesting that replacing saturated fats with linoleic-acid-rich unsaturated fats may improve cardiovascular outcomes.[^3] Because of this, our project focuses on whether higher dietary linoleic acid intake is associated with obesity and cardiovascular risk markers, rather than assuming it is directly harmful.

This topic is important because many Americans may not know how much linoleic acid they consume through fast-food and ultra-processed diets. Over the 20th century, the U.S. diet shifted away from traditional animal fats such as butter and lard toward greater use of industrial vegetable oils, which increased dietary exposure to omega-6 fatty acids.[^4] Omega-6 fatty acids are essential in the diet, but some researchers argue that a high omega-6 to omega-3 ratio may contribute to inflammation-related health concerns.[^4] Omega-3 fatty acids, which are found in foods such as salmon, sardines, tuna, chia seeds, and flaxseeds, may help balance inflammatory processes in the body.[^4] This background helps explain why linoleic acid intake is debated and why it is useful to study its relationship with measurable health outcomes.

For our project, we will focus on obesity and cardiovascular risk because these outcomes can be measured using NHANES variables. BMI will be used as a marker of obesity risk, with BMI of 30 or higher commonly used to classify obesity. Total cholesterol will be used as a cardiovascular risk marker because high cholesterol is associated with greater risk of heart disease. However, these measures have limitations: BMI does not fully capture body composition, and total cholesterol alone does not fully diagnose cardiovascular disease. Therefore, our analysis will examine whether higher linoleic acid intake is associated with BMI and total cholesterol, while clearly stating that the results do not prove causation.

## Hypothesis


We hypothesize that higher dietary linoleic acid intake, commonly found in fast-food and ultra-processed diets that use seed oils, will be associated with higher obesity and cardiovascular risk among U.S. adults. More specifically, we predict that adults with higher linoleic acid intake will have higher average BMI, higher obesity prevalence, and higher total cholesterol levels compared to adults with lower linoleic acid intake. This prediction is based on prior research showing that ultra-processed food consumption is associated with obesity and cardiovascular disease risk, while our project uses linoleic acid intake as a measurable proxy for seed-oil-rich dietary patterns.

### Dataset 1 - Day 1 Dietary Nutrient Intake

The `DR1TOT_L` dataset contains Day 1 dietary recall data from NHANES August 2021–August 2023. Each row represents one participant who completed a 24-hour dietary recall interview, and each column represents a nutrient or dietary measure. This dataset has 8,860 observations and 168 variables.

The most important variable for our project is `DR1TP182`, which measures PFA 18:2 octadecadienoic acid intake in grams. We will use this as our measure of linoleic acid intake because linoleic acid is commonly found in seed oils used in fast-food and ultra-processed diets. We will also use `DR1TKCAL`, which measures total calorie intake in kilocalories, and `DR1TTFAT`, which measures total fat intake in grams. These variables help us understand whether higher linoleic acid intake is related to overall diet size and fat consumption.

A major limitation is that this dataset does not directly measure “seed oil intake” or identify the exact oils used in the foods participants ate. The dietary information is also based on only one 24-hour recall, so it may not fully represent a participant’s long-term eating pattern. Because of this, we will interpret linoleic acid intake as a proxy for seed-oil-rich diets and focus on associations rather than causation.

```python
# Dataset 1 - Day 1 Dietary Nutrient Intake

# Keep only variables relevant to our project
diet_clean = diet[["SEQN", "DR1TP182", "DR1TKCAL", "DR1TTFAT"]].copy()

# Rename columns for readability
diet_clean = diet_clean.rename(columns={
    "DR1TP182": "linoleic_acid_g",
    "DR1TKCAL": "calories_kcal",
    "DR1TTFAT": "total_fat_g"
})

# Demonstrate size of dataset
print("Original diet dataset shape:", diet.shape)
print("Selected diet dataset shape:", diet_clean.shape)

# Check that each row is one participant
print("Duplicate participant IDs:", diet_clean["SEQN"].duplicated().sum())

# Check missing values
missing_table = pd.DataFrame({
    "missing_count": diet_clean.isna().sum(),
    "missing_percent": diet_clean.isna().mean() * 100
})

display(missing_table)

# Summary statistics to check ranges and possible outliers
display(diet_clean.describe())

# Flag suspicious values
suspicious_diet = diet_clean[
    (diet_clean["linoleic_acid_g"] < 0) |
    (diet_clean["calories_kcal"] <= 0) |
    (diet_clean["total_fat_g"] < 0)
]

print("Number of suspicious entries:", suspicious_diet.shape[0])
display(suspicious_diet.head())

# Clean the data by dropping rows missing required dietary variables
diet_clean = diet_clean.dropna(subset=[
    "linoleic_acid_g",
    "calories_kcal",
    "total_fat_g"
])

# Remove impossible values
diet_clean = diet_clean[
    (diet_clean["linoleic_acid_g"] >= 0) &
    (diet_clean["calories_kcal"] > 0) &
    (diet_clean["total_fat_g"] >= 0)
].copy()

# Save cleaned Dataset #1
diet_clean.to_csv(processed_path / "diet_clean.csv", index=False)

print("Final cleaned diet dataset shape:", diet_clean.shape)
display(diet_clean.head())
```

The original dietary dataset had 8,860 rows and 168 variables. After selecting the variables needed for our project, the dataset had 8,860 rows and 4 variables: participant ID, linoleic acid intake, total calories, and total fat. There were no duplicate participant IDs, so the dataset is tidy because each row represents one participant. About 24.45% of rows were missing the main dietary variables, likely because those participants did not have complete Day 1 dietary recall data. We dropped those rows because linoleic acid, calories, and total fat are required for our analysis. After cleaning, the dietary dataset had 6,694 observations and 4 variables.

### Dataset 2 - Body Measures

The `BMX_L` dataset contains body measurement data from NHANES August 2021–August 2023. Each row represents one participant, and each column represents a body measurement collected during the NHANES physical examination. This dataset has 8,860 observations and 22 variables.

The most important variable for our project is `BMXBMI`, which measures Body Mass Index, or BMI, in kg/m². BMI is commonly used to classify adult weight status. A BMI of 30 or higher is commonly used to classify obesity, so this variable will help us measure obesity prevalence in our project.

A limitation of BMI is that it does not fully measure body composition. For example, it does not distinguish between fat mass and muscle mass. However, BMI is still useful for our project because it is a common population-level marker of obesity risk.

```python
# Dataset #2: Body Measures

# Keep only variables relevant to our project
bmx_clean = bmx[["SEQN", "BMXBMI"]].copy()

# Rename columns for readability
bmx_clean = bmx_clean.rename(columns={
    "BMXBMI": "bmi"
})

# Demonstrate size of dataset
print("Original body measures dataset shape:", bmx.shape)
print("Selected body measures dataset shape:", bmx_clean.shape)

# Check that each row is one participant
print("Duplicate participant IDs:", bmx_clean["SEQN"].duplicated().sum())

# Check missing values
missing_table_bmx = pd.DataFrame({
    "missing_count": bmx_clean.isna().sum(),
    "missing_percent": bmx_clean.isna().mean() * 100
})

display(missing_table_bmx)

# Summary statistics to check ranges and possible outliers
display(bmx_clean.describe())

# Flag suspicious BMI values
suspicious_bmx = bmx_clean[
    (bmx_clean["bmi"] <= 0) |
    (bmx_clean["bmi"] > 100)
]

print("Number of suspicious entries:", suspicious_bmx.shape[0])
display(suspicious_bmx.head())

# Clean the data by dropping rows missing BMI
bmx_clean = bmx_clean.dropna(subset=["bmi"])

# Remove impossible or extreme suspicious values
bmx_clean = bmx_clean[
    (bmx_clean["bmi"] > 0) &
    (bmx_clean["bmi"] <= 100)
].copy()

# Create obesity variable
bmx_clean["obese"] = bmx_clean["bmi"] >= 30

# Save cleaned Dataset #2
bmx_clean.to_csv(processed_path / "bmx_clean.csv", index=False)

print("Final cleaned body measures dataset shape:", bmx_clean.shape)
display(bmx_clean.head())
```

The original body measures dataset had 8,860 rows and 22 variables. After selecting the variables needed for our project, the dataset had 8,860 rows and 2 variables: participant ID and BMI. There were no duplicate participant IDs, so the dataset is tidy because each row represents one participant. BMI was missing for 389 participants, which is about 4.39% of the dataset, so we dropped those rows because BMI is required to measure obesity risk. There were no suspicious BMI values based on our cutoff, and the final cleaned body measures dataset had 8,471 observations and 3 variables after adding the obesity indicator.

### Dataset 3 - Total Cholesterol

The `TCHOL_L` dataset contains total cholesterol lab data from NHANES August 2021–August 2023. Each row represents one participant, and each column represents a cholesterol-related lab variable. This dataset has 8,068 observations and 4 variables.

The most important variable for our project is `LBXTC`, which measures total cholesterol in mg/dL. Total cholesterol is a common cardiovascular risk marker. In general, total cholesterol below 200 mg/dL is considered desirable, while 200 mg/dL or higher may indicate higher cardiovascular risk. We will use this variable to measure cardiovascular risk in relation to linoleic acid intake.

A limitation of this dataset is that total cholesterol alone does not fully measure cardiovascular disease risk. Other factors such as HDL cholesterol, LDL cholesterol, triglycerides, blood pressure, smoking, and diabetes status also affect cardiovascular risk. Because of this, we will describe total cholesterol as a cardiovascular risk marker, not as a complete diagnosis of cardiovascular disease.

```python
# Dataset 3: Total Cholesterol

# Keep only variables relevant to our project
chol_clean = chol[["SEQN", "LBXTC"]].copy()

# Rename columns for readability
chol_clean = chol_clean.rename(columns={
    "LBXTC": "total_cholesterol_mg_dl"
})

# Demonstrate size of dataset
print("Original cholesterol dataset shape:", chol.shape)
print("Selected cholesterol dataset shape:", chol_clean.shape)

# Check that each row is one participant
print("Duplicate participant IDs:", chol_clean["SEQN"].duplicated().sum())

# Check missing values
missing_table_chol = pd.DataFrame({
    "missing_count": chol_clean.isna().sum(),
    "missing_percent": chol_clean.isna().mean() * 100
})

display(missing_table_chol)

# Summary statistics to check ranges and possible outliers
display(chol_clean.describe())

# Flag suspicious cholesterol values
suspicious_chol = chol_clean[
    (chol_clean["total_cholesterol_mg_dl"] <= 0) |
    (chol_clean["total_cholesterol_mg_dl"] > 500)
]

print("Number of suspicious entries:", suspicious_chol.shape[0])
display(suspicious_chol.head())

# Clean the data by dropping rows missing total cholesterol
chol_clean = chol_clean.dropna(subset=["total_cholesterol_mg_dl"])

# Remove impossible or extreme suspicious values
chol_clean = chol_clean[
    (chol_clean["total_cholesterol_mg_dl"] > 0) &
    (chol_clean["total_cholesterol_mg_dl"] <= 500)
].copy()

# Create high cholesterol variable
chol_clean["high_cholesterol"] = chol_clean["total_cholesterol_mg_dl"] >= 200

# Save cleaned Dataset #3
chol_clean.to_csv(processed_path / "chol_clean.csv", index=False)

print("Final cleaned cholesterol dataset shape:", chol_clean.shape)
display(chol_clean.head())
```

The original cholesterol dataset had 8,068 rows and 4 variables. After selecting the variables needed for our project, the dataset had 8,068 rows and 2 variables: participant ID and total cholesterol. There were no duplicate participant IDs, so the dataset is tidy because each row represents one participant. Total cholesterol was missing for 1,178 participants, which is about 14.60% of the dataset, so we dropped those rows because cholesterol is required for our cardiovascular risk analysis. There were no suspicious cholesterol values based on our cutoff, and the final cleaned cholesterol dataset had 6,890 observations and 3 variables after adding the high cholesterol indicator.

### Dataset 4 - Demographics

The `DEMO_L` dataset contains demographic information from NHANES August 2021–August 2023. Each row represents one participant, and each column represents a demographic or survey-design variable. This dataset has 11,933 observations and 27 variables.

The most important variables for our project are `RIDAGEYR`, which measures age in years; `RIAGENDR`, which represents sex; `RIDRETH3`, which represents race/ethnicity; and `INDFMPIR`, which represents family income-to-poverty ratio. These variables are important because age, sex, race/ethnicity, and income may affect both diet and health outcomes. We will also use age to filter the dataset to adults.

A limitation of this dataset is that it does not include every possible factor that may influence BMI or cholesterol. For example, physical activity, smoking, medication use, and medical history may also affect obesity and cardiovascular risk. Some demographic variables may also be missing or grouped broadly to protect participant privacy.

```python
# Dataset 4: Demographics

# Keep only variables relevant to our project
demo_clean = demo[["SEQN", "RIDAGEYR", "RIAGENDR", "RIDRETH3", "INDFMPIR"]].copy()

# Rename columns for readability
demo_clean = demo_clean.rename(columns={
    "RIDAGEYR": "age",
    "RIAGENDR": "sex",
    "RIDRETH3": "race_ethnicity",
    "INDFMPIR": "income_poverty_ratio"
})

# Demonstrate size of dataset
print("Original demographics dataset shape:", demo.shape)
print("Selected demographics dataset shape:", demo_clean.shape)

# Check that each row is one participant
print("Duplicate participant IDs:", demo_clean["SEQN"].duplicated().sum())

# Check missing values
missing_table_demo = pd.DataFrame({
    "missing_count": demo_clean.isna().sum(),
    "missing_percent": demo_clean.isna().mean() * 100
})

display(missing_table_demo)

# Summary statistics to check ranges and possible outliers
display(demo_clean.describe())

# Flag suspicious demographic values
suspicious_demo = demo_clean[
    (demo_clean["age"] < 0) |
    (~demo_clean["sex"].isin([1, 2])) |
    (~demo_clean["race_ethnicity"].isin([1, 2, 3, 4, 6, 7])) |
    (demo_clean["income_poverty_ratio"] < 0)
]

print("Number of suspicious entries:", suspicious_demo.shape[0])
display(suspicious_demo.head())

# Clean the data:
# Keep adults only because our research question focuses on U.S. adults.
demo_clean = demo_clean[demo_clean["age"] >= 20].copy()

# Drop rows missing required demographic variables
demo_clean = demo_clean.dropna(subset=[
    "age",
    "sex",
    "race_ethnicity"
])

# Income-to-poverty ratio is useful but may be missing, so we keep missing income values for now.
# We may drop or handle income missingness later depending on the final model.

# Remove impossible demographic values
demo_clean = demo_clean[
    (demo_clean["age"] >= 20) &
    (demo_clean["sex"].isin([1, 2])) &
    (demo_clean["race_ethnicity"].isin([1, 2, 3, 4, 6, 7]))
].copy()

# Save cleaned Dataset #4
demo_clean.to_csv(processed_path / "demo_clean.csv", index=False)

print("Final cleaned demographics dataset shape:", demo_clean.shape)
display(demo_clean.head())
```

The original demographics dataset had 11,933 rows and 27 variables. After selecting the variables needed for our project, the dataset had 11,933 rows and 5 variables: participant ID, age, sex, race/ethnicity, and income-to-poverty ratio. There were no duplicate participant IDs, so the dataset is tidy because each row represents one participant. Age, sex, and race/ethnicity had no missing values, but income-to-poverty ratio was missing for 2,041 participants, or about 17.10% of the dataset. We kept missing income values for now because income is useful but not required for the main analysis, then filtered the dataset to adults age 20 and older. The final cleaned demographics dataset had 7,809 observations and 5 variables.
