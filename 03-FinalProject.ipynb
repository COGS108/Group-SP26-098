## Abstract

This project investigates whether higher dietary linoleic acid intake is associated with obesity and cardiovascular risk among U.S. adults. Linoleic acid is an omega-6 fatty acid commonly found in seed oils used in many fast-food and ultra-processed foods. Using NHANES 2021–2023 data, we analyzed dietary intake, BMI, total cholesterol, and demographic variables to examine whether linoleic acid intake is related to higher BMI and cholesterol levels.

We conducted exploratory data analysis by comparing low, medium, and high linoleic acid intake groups across BMI, obesity prevalence, total cholesterol, and high cholesterol prevalence. We also examined age, calorie intake, total fat intake, sex, race/ethnicity, and income patterns to understand possible confounding variables. Finally, we used regression models to test whether linoleic acid intake was associated with BMI or total cholesterol after controlling for calories, total fat, age, sex, race/ethnicity, and income-to-poverty ratio.

The results did not strongly support our hypothesis. BMI, obesity prevalence, total cholesterol, and high cholesterol prevalence were fairly similar across linoleic acid intake groups. Regression results also showed that linoleic acid intake was not significantly associated with BMI or total cholesterol after controlling for other variables. These findings suggest that linoleic acid intake alone may not be a strong predictor of obesity or cholesterol in this sample, and that broader diet quality may matter more than focusing on one nutrient by itself.

## Research Question

Is higher dietary linoleic acid intake, used as a proxy for seed-oil-rich diets, associated with higher BMI and total cholesterol levels among U.S. adults?

## Background and Prior Work

Dietary patterns in the United States have shifted toward greater consumption of ultra-processed foods, many of which are energy-dense and contain refined vegetable oils. Linoleic acid is an omega-6 polyunsaturated fatty acid commonly found in oils such as corn, sunflower, soybean, and canola oil.[^1] According to the CDC’s National Center for Health Statistics, ultra-processed foods made up about 55% of total daily calorie intake among people age 1 and older in the United States from August 2021 to August 2023, with an even higher percentage among youth.[^1] Prior research has also found that higher consumption of ultra-processed foods is associated with increased risk of obesity, type 2 diabetes, and cardiovascular disease.[^2] However, research on linoleic acid itself is more mixed, with some evidence suggesting that replacing saturated fats with linoleic-acid-rich unsaturated fats may improve cardiovascular outcomes.[^3] Because of this, our project focuses on whether higher dietary linoleic acid intake is associated with obesity and cardiovascular risk markers, rather than assuming it is directly harmful.

This topic is important because many Americans may not know how much linoleic acid they consume through fast-food and ultra-processed diets. Over the 20th century, the U.S. diet shifted away from traditional animal fats such as butter and lard toward greater use of industrial vegetable oils, which increased dietary exposure to omega-6 fatty acids.[^4] Omega-6 fatty acids are essential in the diet, but some researchers argue that a high omega-6 to omega-3 ratio may contribute to inflammation-related health concerns.[^4] Omega-3 fatty acids, which are found in foods such as salmon, sardines, tuna, chia seeds, and flaxseeds, may help balance inflammatory processes in the body.[^4] This background helps explain why linoleic acid intake is debated and why it is useful to study its relationship with measurable health outcomes.

For our project, we will focus on obesity and cardiovascular risk because these outcomes can be measured using NHANES variables. BMI will be used as a marker of obesity risk, with BMI of 30 or higher commonly used to classify obesity. Total cholesterol will be used as a cardiovascular risk marker because high cholesterol is associated with greater risk of heart disease. However, these measures have limitations: BMI does not fully capture body composition, and total cholesterol alone does not fully diagnose cardiovascular disease. Therefore, our analysis will examine whether higher linoleic acid intake is associated with BMI and total cholesterol, while clearly stating that the results do not prove causation.

## Hypothesis


We hypothesize that higher dietary linoleic acid intake, commonly found in fast-food and ultra-processed diets that use seed oils, will be associated with higher obesity and cardiovascular risk among U.S. adults. More specifically, we predict that adults with higher linoleic acid intake will have higher average BMI, higher obesity prevalence, and higher total cholesterol levels compared to adults with lower linoleic acid intake. This prediction is based on prior research showing that ultra-processed food consumption is associated with obesity and cardiovascular disease risk, while our project uses linoleic acid intake as a measurable proxy for seed-oil-rich dietary patterns.

## Data

### Data Overview

We are using four NHANES August 2021–August 2023 datasets from the CDC/National Center for Health Statistics. These datasets are publicly available as XPT files and do not require special permission. We combined them using `SEQN`, the respondent sequence number, so that each row in our final dataset represents one adult participant.

**Dataset #1**
- **Dataset Name:** Dietary Interview - Total Nutrient Intakes, First Day (`DR1TOT_L`)
- **Link to the dataset:** https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/DR1TOT_L.htm
- **Number of observations:** 8,860
- **Number of variables:** 168
- **Relevant variables:** `SEQN` for participant ID, `DR1TP182` for linoleic acid intake in grams, `DR1TKCAL` for total calories in kcal, and `DR1TTFAT` for total fat in grams.
- **Shortcomings:** This dataset uses a 24-hour dietary recall, so one day may not represent a participant’s usual diet. It also does not directly identify seed oil or fast-food intake, so we use linoleic acid intake as a proxy.

**Dataset #2**
- **Dataset Name:** Body Measures (`BMX_L`)
- **Link to the dataset:** https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/BMX_L.htm
- **Number of observations:** 8,860
- **Number of variables:** 22
- **Relevant variables:** `SEQN` for participant ID and `BMXBMI` for Body Mass Index in kg/m².
- **Shortcomings:** BMI is useful for measuring obesity risk, but it does not fully describe body composition or overall health.

**Dataset #3**
- **Dataset Name:** Cholesterol - Total (`TCHOL_L`)
- **Link to the dataset:** https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/TCHOL_L.htm
- **Number of observations:** 8,068
- **Number of variables:** 4
- **Relevant variables:** `SEQN` for participant ID and `LBXTC` for total cholesterol in mg/dL.
- **Shortcomings:** Total cholesterol is only one cardiovascular risk marker. It does not fully measure cardiovascular disease risk by itself.

**Dataset #4**
- **Dataset Name:** Demographic Variables and Sample Weights (`DEMO_L`)
- **Link to the dataset:** https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/DEMO_L.htm
- **Number of observations:** 11,933
- **Number of variables:** 27
- **Relevant variables:** `SEQN` for participant ID, `RIDAGEYR` for age, `RIAGENDR` for sex, `RIDRETH3` for race/ethnicity, and `INDFMPIR` for income-to-poverty ratio.
- **Shortcomings:** Some demographic variables are limited for privacy reasons, and this dataset does not include every possible lifestyle factor that may affect BMI or cholesterol.

We merged these datasets using `SEQN`, filtered the data to adults, removed missing values in the main analysis variables, and created new variables for obesity status and high cholesterol status.

### Merged NHANES Analysis Dataset

The full data wrangling process was completed across the previous checkpoints. In `01-DataCheckpoint.ipynb`, we cleaned the NHANES dietary intake, body measures, total cholesterol, and demographic datasets separately. In `02-EDACheckpoint.ipynb`, these cleaned datasets were merged using `SEQN`, the participant ID, and saved as a processed merged dataset. For the final project, we load this processed dataset from `data/02-processed` and use it for EDA and regression analysis.

The final merged dataset includes linoleic acid intake, total calories, total fat, BMI, obesity status, total cholesterol, high cholesterol status, age, sex, race/ethnicity, and income-to-poverty ratio. Each row represents one adult participant. We also create a `linoleic_group` variable that divides participants into low, medium, and high linoleic acid intake groups. This allows us to compare health outcomes across intake groups while also using dietary and demographic variables in the regression analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from pathlib import Path

processed_path = Path("data/02-processed")
results_path = Path("results")
results_path.mkdir(exist_ok=True)

df = pd.read_csv(processed_path / "merged_eda_dataset.csv")

df["linoleic_group"] = pd.qcut(
    df["linoleic_acid_g"],
    q=3,
    labels=["Low", "Medium", "High"]
)

print("Processed dataset shape:", df.shape)
print("Duplicate participant IDs:", df["SEQN"].duplicated().sum())

display(df.head())
```

## Results

### Exploratory Data Analysis

The full EDA process was completed in `02-EDACheckpoint.ipynb`. For the final report, we focus only on the EDA results most relevant to our research question. We examine whether linoleic acid intake differs meaningfully across health-risk markers by comparing low, medium, and high linoleic acid intake groups.

Our EDA looks at BMI, obesity prevalence, total cholesterol, and high cholesterol prevalence across these intake groups. We also examine age, calorie intake, total fat intake, and demographic patterns to better understand possible confounding variables before moving into regression analysis. These EDA results help show the initial patterns in the data and explain why the regression models control for dietary and demographic variables.

```python
eda_summary = df.groupby("linoleic_group", observed=False).agg({
    "bmi": "mean",
    "obese": "mean",
    "total_cholesterol_mg_dl": "mean",
    "high_cholesterol": "mean"
})

eda_summary["obese"] *= 100
eda_summary["high_cholesterol"] *= 100

eda_summary = eda_summary.rename(columns={
    "bmi": "Average BMI",
    "obese": "Obesity Prevalence (%)",
    "total_cholesterol_mg_dl": "Average Total Cholesterol (mg/dL)",
    "high_cholesterol": "High Cholesterol Prevalence (%)"
})

display(eda_summary.round(2))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

df.boxplot(column="bmi", by="linoleic_group", ax=axes[0])
df.boxplot(column="total_cholesterol_mg_dl", by="linoleic_group", ax=axes[1])

axes[0].set(title="BMI by Linoleic Acid Group", xlabel="Linoleic Acid Group", ylabel="BMI")
axes[1].set(title="Total Cholesterol by Linoleic Acid Group", xlabel="Linoleic Acid Group", ylabel="Total Cholesterol (mg/dL)")

fig.suptitle("")
plt.tight_layout()
plt.savefig(results_path / "boxplots_health_markers.png", dpi=300, bbox_inches="tight")
plt.show()

eda_summary[[
    "Obesity Prevalence (%)",
    "High Cholesterol Prevalence (%)"
]].plot(kind="bar", figsize=(9, 5))

plt.title("Health Risk Prevalence by Linoleic Acid Group")
plt.xlabel("Linoleic Acid Group")
plt.ylabel("Percent")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(results_path / "prevalence_by_linoleic_group.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Deeper EDA: Age and Diet-Related Variables

Next, we examine how age, calorie intake, and total fat intake differ across low, medium, and high linoleic acid intake groups. This helps us understand whether linoleic acid intake is connected to other variables before running regression analysis.

```python
distribution_summary = df.groupby("linoleic_group", observed=False)[[
    "age",
    "calories_kcal",
    "total_fat_g",
    "linoleic_acid_g"
]].mean()

distribution_summary = distribution_summary.rename(columns={
    "age": "Average Age",
    "calories_kcal": "Average Calories",
    "total_fat_g": "Average Total Fat (g)",
    "linoleic_acid_g": "Average Linoleic Acid (g)"
})

display(distribution_summary.round(2))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

df.boxplot(column="age", by="linoleic_group", ax=axes[0])
df.boxplot(column="calories_kcal", by="linoleic_group", ax=axes[1])
df.boxplot(column="total_fat_g", by="linoleic_group", ax=axes[2])

axes[0].set(title="Age by Linoleic Acid Group", xlabel="Linoleic Acid Group", ylabel="Age")
axes[1].set(title="Calories by Linoleic Acid Group", xlabel="Linoleic Acid Group", ylabel="Calories")
axes[2].set(title="Total Fat by Linoleic Acid Group", xlabel="Linoleic Acid Group", ylabel="Total Fat (g)")

fig.suptitle("")
plt.tight_layout()
plt.savefig(results_path / "age_calories_fat_by_linoleic_group.png", dpi=300, bbox_inches="tight")
plt.show()
```

The deeper EDA shows that participants with higher linoleic acid intake also tend to consume more calories and total fat. This suggests that raw linoleic acid intake may partly reflect overall food intake, not only seed-oil-rich dietary patterns. Because of this, the regression analysis controls for calories, total fat, and age.

### Demographic EDA: Linoleic Acid Intake Across Groups

Next, we examine whether average linoleic acid intake differs across sex, race/ethnicity, and income groups. This helps us understand whether demographic patterns may affect dietary intake before moving into regression analysis.

```python
df_demo = df.copy()

df_demo["sex_label"] = df_demo["sex"].map({1: "Male", 2: "Female"})

df_demo["race_ethnicity_label"] = df_demo["race_ethnicity"].map({
    1: "Mexican American",
    2: "Other Hispanic",
    3: "Non-Hispanic White",
    4: "Non-Hispanic Black",
    6: "Non-Hispanic Asian",
    7: "Other/Multi-Racial"
})

df_demo["income_group"] = pd.cut(
    df_demo["income_poverty_ratio"],
    bins=[0, 1.3, 3.5, 5],
    labels=["Low Income", "Middle Income", "High Income"],
    include_lowest=True
)

sex_summary = df_demo.groupby("sex_label", observed=False)["linoleic_acid_g"].mean()
race_summary = df_demo.groupby("race_ethnicity_label", observed=False)["linoleic_acid_g"].mean().sort_values()
income_summary = df_demo.groupby("income_group", observed=False)["linoleic_acid_g"].mean()

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

sex_summary.plot(kind="bar", ax=axes[0])
axes[0].set(title="Linoleic Acid Intake by Sex", xlabel="Sex", ylabel="Average Intake (g)")
axes[0].tick_params(axis="x", rotation=0)

race_summary.plot(kind="barh", ax=axes[1])
axes[1].set(title="Linoleic Acid Intake by Race/Ethnicity", xlabel="Average Intake (g)", ylabel="")

income_summary.plot(kind="bar", ax=axes[2])
axes[2].set(title="Linoleic Acid Intake by Income Group", xlabel="Income Group", ylabel="Average Intake (g)")
axes[2].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.savefig(results_path / "demographic_linoleic_intake.png", dpi=300, bbox_inches="tight")
plt.show()
```

The demographic EDA shows that linoleic acid intake differs somewhat across sex, race/ethnicity, and income groups. These differences suggest that diet patterns may vary across demographic groups. Because demographic factors may affect both dietary intake and health outcomes, the regression analysis controls for sex, race/ethnicity, and income-to-poverty ratio.

### Regression Analysis

To move beyond the EDA, we use regression models to test whether dietary linoleic acid intake is associated with our main health-risk markers after accounting for other factors. We use BMI and total cholesterol as continuous outcome variables. Linoleic acid intake is the main predictor, and we control for total calories, total fat, age, sex, race/ethnicity, and income-to-poverty ratio.

These control variables are important because the EDA showed that higher linoleic acid intake is closely related to higher total calorie intake and higher total fat intake. Demographic variables may also affect both diet patterns and health outcomes. By including these controls, the regression models help test whether linoleic acid intake has an association with BMI or cholesterol beyond general food intake and demographic differences.

Because NHANES is observational data, the results can show association but cannot prove causation.

```python
model_df = df.dropna(subset=[
    "linoleic_acid_g",
    "calories_kcal",
    "total_fat_g",
    "bmi",
    "total_cholesterol_mg_dl",
    "age",
    "sex",
    "race_ethnicity",
    "income_poverty_ratio"
]).copy()

controls = """
linoleic_acid_g + calories_kcal + total_fat_g + age
+ C(sex) + C(race_ethnicity) + income_poverty_ratio
"""

bmi_model = smf.ols(f"bmi ~ {controls}", data=model_df).fit()
chol_model = smf.ols(f"total_cholesterol_mg_dl ~ {controls}", data=model_df).fit()

regression_results = pd.DataFrame({
    "Outcome": ["BMI", "Total Cholesterol"],
    "Linoleic Acid Coefficient": [
        bmi_model.params["linoleic_acid_g"],
        chol_model.params["linoleic_acid_g"]
    ],
    "P-value": [
        bmi_model.pvalues["linoleic_acid_g"],
        chol_model.pvalues["linoleic_acid_g"]
    ],
    "R-squared": [
        bmi_model.rsquared,
        chol_model.rsquared
    ]
})

display(regression_results.round(4))
```

### Regression Results Interpretation

The regression results show that linoleic acid intake is not significantly associated with either BMI or total cholesterol after controlling for calories, total fat, age, sex, race/ethnicity, and income-to-poverty ratio. For BMI, the linoleic acid coefficient is -0.0063 with a p-value of 0.7098, meaning the relationship is very small and not statistically significant. For total cholesterol, the coefficient is 0.0293 with a p-value of 0.7685, which is also very small and not statistically significant.

The R-squared values are also low, around 0.03 for both models. This means that the variables in these models explain only a small portion of the variation in BMI and total cholesterol. Overall, the regression results do not support the hypothesis that higher linoleic acid intake is associated with higher BMI or higher total cholesterol in this sample.

## Study Summary and Findings

In this project, we investigated whether higher dietary linoleic acid intake is associated with obesity and cardiovascular risk among U.S. adults. We used data from NHANES, the National Health and Nutrition Examination Survey, and conducted exploratory data analysis followed by regression modeling. BMI and obesity status were used as obesity-related outcomes, while total cholesterol and high cholesterol status were used as cardiovascular risk markers.

Our hypothesis was that higher dietary linoleic acid intake, commonly found in seed-oil-rich fast-food and ultra-processed diets, would be associated with higher BMI, higher obesity prevalence, and higher total cholesterol levels. This hypothesis was based on prior research showing that ultra-processed food consumption is associated with obesity and cardiovascular disease risk. In our project, linoleic acid intake was used as a measurable proxy for seed-oil-rich dietary patterns.

However, our results did not strongly support the hypothesis. The EDA showed that BMI, obesity prevalence, total cholesterol, and high cholesterol prevalence were fairly similar across low, medium, and high linoleic acid intake groups. The regression models also showed that linoleic acid intake was not significantly associated with BMI or total cholesterol after controlling for calories, total fat, age, sex, race/ethnicity, and income-to-poverty ratio. This suggests that while ultra-processed diets may be linked to poor health outcomes, isolating one nutrient such as linoleic acid produces a weaker and more complicated relationship.

## Limitations

There are several limitations to this study. First, NHANES dietary data relies on 24-hour dietary recalls, which may not represent a person’s usual long-term diet. A participant’s intake on one day may be different from their normal eating habits, and self-reported diet data can include recall errors.

Second, NHANES does not directly measure seed oil intake or fast-food oil intake. Because of this, we used linoleic acid intake as a proxy for seed-oil-rich diets. This is useful, but imperfect, because linoleic acid can come from many foods and does not identify the exact source of the oil or the overall quality of the diet.

Third, BMI and total cholesterol are limited health measures. BMI does not distinguish between muscle mass and body fat, so it does not fully describe body composition. Total cholesterol also does not show the full cardiovascular risk picture because it does not separate HDL, LDL, triglycerides, blood pressure, diabetes status, or medication use. Because of these limitations, the relationship between linoleic acid intake and cardiovascular health cannot be fully studied using only BMI, total cholesterol, and one day of dietary recall.

## Future Work

Future research should use more detailed dietary data, including multiple days of dietary recall or direct measures of ultra-processed food and seed oil consumption. It would also be useful to include additional health markers such as LDL cholesterol, HDL cholesterol, triglycerides, blood pressure, diabetes status, physical activity, and medication use. Longitudinal data would also help determine whether long-term dietary patterns are related to changes in BMI or cardiovascular risk over time.

Even though our results did not show a strong relationship between linoleic acid intake alone and BMI or total cholesterol, the topic remains important. Many fast-food and ultra-processed foods contain seed oils, but they also contain many other factors that may affect health, such as high calories, sodium, added sugars, and low fiber. The bigger takeaway from this project is that overall diet quality may matter more than focusing on one nutrient by itself.
