## Research Question

To what extent is the dietary consumption of ultra-processed foods that contain seed oils, particularly those rich in linoleic-acid, associated with health risks among the U.S. population?

## Background and Prior Work

The dietary patterns in the U.S. have greatly shifted towards a greater consumption of ultra-processed foods that often incorporate seeds oils, predeominanlty linoleic acid, which is an omega-6 polyunsaturated fatty acid commonly found in corn, sunflower, and canola oils. [^1] The CDC's National Center for Health Statistics states that ultra-processed foods constitute about 55% of total daily calorie intake in Americans that are 1 year old and older; this percentage is greater in American youth populations.[^1] Furthermore, other research has observed that an increased consumption of ultra-processed foods can lead to increased health risks of type 2-diabetes, obesity, and cardiovascular disease. [^2] On the contrary, there is evidence to suggest that linoleic acid itself does not always increase such health risks and can actually improve cardiovascular outcomes as a replacement for saturated fats. [^3] Such findings reveal that overconsumption of unltra-processed foods is likely more important than that of seed oils.
	The average American is not aware of the presence of seed oils. Through the 20th century, the US shifted from consuming and using butters and lard to industrial seed oils [^4].  The over consumption of seed oils has caused an imbalance in omega-3 / omega-6 fatty acids. Omega 6 fatty acids are very important to the human diet, despite its importance, when over consumed, it can produce archadonic acid which can ultimatley cause inflamation [^5]. Omega 3 fatty acids play a key role in lowering those inflammation levels [^5]. Omega 3 fatty acids can be found in foods like tuna, sardines, salmon, and seeds like chia seeds [^5]. In addition to the inflammation concerns of omega 6 fatty acids, many of them are consumed afteer heating. The heating of omega 6 fatty acids causes a release of aldehydes, trans fats, and polymers which are all sevrley toxic to the human body [^4]. It is critical to inform and understand the dangers of the over consumption of seed oils and lineoelic acids in order to preserve the health of the American people.

## Hypothesis


We hypothesize that increased consumption of foods likely containing seed oils and linoleic acids, approximated via ultra-processed food intake, is associated with a higher health risk in the American population. More specifically, we predict that groups with higher consumption of ultra-processed foods will show consistent patterns with higher risk, such as greater prevalence of bad dietary profiles across income and age groups. This is based on previous research suggesting the fact that ultra-processed foods, often carrying seed oils, are related to increased risk of metabolic disorder, cardiovascular disease, and obesity. Hence, we expect to see that populations with higher ultra-processed food consumption are more at risk. 

## Data

The ideal dataset would include individual-level dietary and health data from a large sample of the American population. Important variables would include seed oil intake, linoleic acid intake, total calories, BMI, total cholesterol, age, sex, race/ethnicity, and income level. We would need several thousand observations, with each row representing one participant. The data would be collected through dietary recall interviews, physical exams, lab tests, and demographic questionnaires, then stored in tables that can be merged using a participant ID.

For our actual project, we plan to use NHANES August 2021–August 2023, which is publicly available from the CDC and does not require special permission. The data can be downloaded as XPT files and imported into Python or R. We would use DR1TOT_L for dietary nutrients, especially DR1T18_2 for linoleic acid intake; BMX_L for BMI; TCHOL_L for total cholesterol; and DEMO_L for demographic variables. Since NHANES does not directly measure seed oil intake, we would use linoleic acid intake as a proxy for diets involving seed oils.
